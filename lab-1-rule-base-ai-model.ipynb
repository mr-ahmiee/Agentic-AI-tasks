```python
def weather_advice(temperature_c, humidity_percent, wind_speed_kmh, aqi):
    """Rule-based weather advisor using simple IF/THEN logic"""
    
    advice = []

    # Rule 1: Temperature
    if temperature_c >= 35:
        advice.append("Extreme heat: stay hydrated and avoid the sun.")
    elif temperature_c <= 5:
        advice.append("Cold weather: wear a warm jacket.")

    # Rule 2: Humidity
    if humidity_percent >= 80:
        advice.append("High humidity: expect possible rain, carry an umbrella.")

    # New Rule 3: Wind speed warning
    if wind_speed_kmh >= 40:
        advice.append("Strong winds: be careful outdoors and secure loose objects.")

    # New Rule 4: Air quality warning
    if aqi >= 150:
        advice.append("Poor air quality: avoid prolonged outdoor activities.")

    # If no special condition exists
    if not advice:
        advice.append("Conditions look moderate. No special precautions needed.")

    return advice


if __name__ == "__main__":

    # Five different test cases
    sample_days = [
        (38, 40, 20, 80),    # Test 1: Extreme heat
        (3, 55, 10, 50),     # Test 2: Cold weather
        (28, 85, 15, 100),   # Test 3: High humidity
        (25, 60, 50, 80),    # Test 4: Strong winds
        (30, 70, 25, 180)    # Test 5: Poor air quality
    ]

    for i, (temp, humidity, wind, aqi) in enumerate(sample_days, 1):
        result = weather_advice(temp, humidity, wind, aqi)

        print(f"\nTest Case {i}")
        print(
            f"Temp={temp}C, "
            f"Humidity={humidity}%, "
            f"Wind={wind} km/h, "
            f"AQI={aqi}"
        )
        print("Advice:")
        
        for message in result:
            print("-", message)
```

```python
# Build a Simple Data-Driven Spam Classifier

def classify_email(email_text, spam_keywords, threshold=2):
    """Classifies an email as Spam or Legitimate using keyword counting."""

    text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in text:
            count += 1

    if count >= threshold:
        label = "Spam"
    else:
        label = "Legitimate"

    return label, count


if __name__ == "__main__":

    # Spam-indicator keywords
    spam_keywords = [
        "free",
        "winner",
        "urgent",
        "click here",
        "prize",
        "offer",
        "congratulations"
    ]

    # Six sample emails with correct labels
    test_emails = [
        ("Congratulations! You are a winner! Claim your free prize now.", "Spam"),

        ("Urgent! Click here to claim your prize and free offer.", "Spam"),

        ("You have won a free gift. Congratulations, winner!", "Spam"),

        ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),

        ("Please find attached the meeting notes from today's class.", "Legitimate"),

        ("Your appointment is confirmed for Monday at 10 AM.", "Legitimate")
    ]

    correct = 0

    print("SPAM CLASSIFIER RESULTS")
    print("-" * 50)

    for i, (email, actual_label) in enumerate(test_emails, 1):

        predicted_label, keyword_count = classify_email(
            email,
            spam_keywords,
            threshold=2
        )

        if predicted_label == actual_label:
            correct += 1

        print(f"\nEmail {i}:")
        print(f"Text: {email}")
        print(f"Keywords found: {keyword_count}")
        print(f"Predicted: {predicted_label}")
        print(f"Actual:    {actual_label}")

    # Calculate accuracy
    accuracy = (correct / len(test_emails)) * 100

    print("\n" + "-" * 50)
    print(f"Correct classifications: {correct}/{len(test_emails)}")
    print(f"Accuracy: {accuracy:.1f}%")
```
